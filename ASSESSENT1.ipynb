2.  Write a Python function that converts temperature from Fahrenheit to Celsius. Include a docstring, type hints, and input validation. Test it with three values including one invalid input.




```python
def fahrenheit_to_celsius(temp_f: float) -> float:
    """Convert temperature from Fahrenheit to Celsius.

    Parameters:
        temp_f: Temperature in Fahrenheit. Must be a numerical value
                and cannot be below Absolute Zero (-459.67°F).

    Returns:
        Temperature in Celsius rounded to two decimal places.
    """
    # 1. Type Validation (Ensures input is a number)
    if not isinstance(temp_f, (int, float)):
        raise TypeError(f"Temperature must be an integer or float. Got: {type(temp_f).__name__}")

    # 2. Value Validation (Ensures input is physically possible)
    if temp_f < -459.67:
        raise ValueError(f"Temperature cannot be below Absolute Zero (-459.67°F). Got: {temp_f}")

    # 3. Calculation & Rounding
    temp_c = (temp_f - 32) * 5 / 9
    return round(temp_c, 2)


# --- Testing the Function ---

# Test Case 1: Standard positive temperature (Boiling point of water)
try:
    print(f"212°F is equal to {fahrenheit_to_celsius(212)}°C")
except (TypeError, ValueError) as e:
    print(f"Test 1 Failed: {e}")

# Test Case 2: Negative temperature (Freezing point blend)
try:
    print(f"-40°F is equal to {fahrenheit_to_celsius(-40)}°C")
except (TypeError, ValueError) as e:
    print(f"Test 2 Failed: {e}")

# Test Case 3: Invalid Input (String type value)
try:
    print("Attempting to convert 'hot'...")
    fahrenheit_to_celsius("hot")
except (TypeError, ValueError) as e:
    print(f"❌ Test 3 Caught Expected Error: {e}")
```

#
# MANUAL VERIFICATION OF BARLOW'S WALL THICKNESS
# Formula:  $$ t_min = (P * D) / (2 * S * F * E) $$
#

```python


# Given Input Parameters
P = 4467      # Pressure (psi)
D = 10.75     # Outer Diameter (inches)
S = 65000     # SMYS / Material Strength (psi)
F = 0.72      # Default Design Factor
E = 1.0       # Default Joint Factor

# Step 1: Calculate the Top Half (Numerator)
# 4467 * 10.75
top_half = P * D
print(f"Step 1: Top Half (Bursting Force) = {top_half}")

# Step 2: Calculate the Bottom Half (Denominator)
# 2 * 65000 * 0.72 * 1.0
bottom_half = 2 * S * F * E
print(f"Step 2: Bottom Half (Material Resistance) = {bottom_half}")

# Step 3: Divide Top by Bottom to get the Raw Answer
raw_t_min = top_half / bottom_half
print(f"Step 3: Raw Unrounded Result = {raw_t_min}")

# Step 4: Round to 4 Decimal Places (matching the function's logic)
final_t_min = round(raw_t_min, 4)
print(f"Step 4: Final Rounded Output = {final_t_min} inches")

# Verification Verification
print(f"\nVerification Success: {final_t_min} matches the function output of 0.513!")
```

# **QUESTION 6**
  Write a Python class called SubseaPipeline with properties for length_km, od_inches, wt_mm, material_grade, and a method that calculates the pipe volume in cubic metres.


```python
import math

class SubseaPipeline:
    def __init__(self, length_km: float, od_inches: float, wt_mm: float, material_grade: str):
        """
        Initialises the SubseaPipeline object.

        Parameters:
            length_km (float): Pipeline length in kilometres.
            od_inches (float): Outer diameter of the pipe in inches.
            wt_mm (float): Wall thickness of the pipe in millimetres.
            material_grade (str): Steel strength rating (e.g., 'X65', 'X70').
        """
        self.length_km = length_km
        self.od_inches = od_inches
        self.wt_mm = wt_mm
        self.material_grade = material_grade

    def calculate_internal_volume(self) -> float:
        """
        Calculates the internal volume of the pipe channel in cubic metres (m³).
        Formula: Volume = pi * (Internal Radius)² * Length
        """
        # 1. Standardise all measurements into Metric Metres
        # Convert total length from kilometres to metres
        length_m = self.length_km * 1000.0

        # Convert outer diameter from inches to metres (1 inch = 0.0254 metres)
        od_m = self.od_inches * 0.0254

        # Convert wall thickness from millimetres to metres (1 mm = 0.001 metres)
        wt_m = self.wt_mm * 0.001

        # 2. Find the internal diameter (ID) by subtracting double the wall thickness
        id_m = od_m - (2.0 * wt_m)

        # 3. Calculate internal radius (r)
        radius_m = id_m / 2.0

        # 4. Apply cylinder volume math: V = pi * r² * L
        volume_m3 = math.pi * (radius_m ** 2) * length_m

        # Return the final result rounded to 2 decimal places
        return round(volume_m3, 2)


# --- Test Execution Block ---

# Create an object with parameters matching standard subsea field setups
# Length = 15km, Outer Diameter = 12 inches, Wall Thickness = 14.3mm, Grade = X65
pipeline_asset = SubseaPipeline(
    length_km=15.0,
    od_inches=12.0,
    wt_mm=14.3,
    material_grade="X65"
)

# Run the method and pull out the cubic contents space
total_vol = pipeline_asset.calculate_internal_volume()

print(f"Asset Grade: {pipeline_asset.material_grade}")
print(f"Pipeline Total Internal Capacity: {total_vol} cubic metres (m³)")
```

Why is version control essential for engineering software? Describe three specific scenarios where Git would prevent costly errors in an offshore engineering context.


 Scenario 1: Preventing the Overwriting of Calibrated Hydrodynamic Parameters

1.   The Risk: An offshore analyst updates an unversioned subsea pipeline design script to change wave-loading constants for a shallow-water project. A few days later, a second engineer opens the shared file network and inadvertently runs the same script for a deepwater riser calculation, completely unaware that the calibration parameters were modified. This calculation error could cause the pipe wall thickness to be critically underestimated, resulting in structural failure under deep ocean pressures.
2.   How Git Prevents It: Git enforces isolation through Branches. The first analyst must create a dedicated workspace (e.g., feature/shallow-water-wave-load) to perform their task. The primary engineering blueprint (main branch) remains perfectly locked and completely unaffected until a rigorous engineering peer review is conducted and approved.

 Scenario 2: Recovering Instantly from a Corrupted Emergency Shut-down Update



1.  The Risk: During an active software update sprint on an offshore platform, an automated control valve script is modified to speed up execution. During live testing, the newly introduced loop triggers an unexpected memory leak, causing the platform’s sensor display board to lock up and freeze. Without version tracking, engineers would have to manually comb through hundreds of lines of code to identify the error, all while the facility remains in an expensive operational downtime window costing hundreds of thousands of dollars per hour.
2.   How Git Prevents It: Git provides instant disaster recovery through git revert or git checkout. Instead of hunting for the bug under extreme pressure, the systems team can execute a single terminal command to roll the operational software environment backward to the last known stable state in under 5 seconds. This instantly restores safe production before debugging begins on a separate machine.

 Scenario 3: Eliminating the Confusion of "Final_v2_edit.py" Disasters

1.   The Risk: A subsea engineering consultancy is validating internal burst calculations using Barlow's formula. Engineers always pull the absolute latest validated script from the main server, eliminating human sorting errors and manufacturing waste.

QUESTION 9
 Explain the try/except pattern. Write a function that safely divides two numbers and returns ‘Error: Division by zero’ if the denominator is zero.


In Python, try/except is used to handle errors without stopping the

1.   try → contains the code that might cause an error.
2.    itemexcept → tells Python what to do if that error happens.


```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Division by zero"

# Test cases
print(f"5 / 2 = {safe_divide(5, 2)}")
print(f"10 / 0 = {safe_divide(10, 0)}")
print(f"-7 / 3 = {safe_divide(-7, 3)}")
```
